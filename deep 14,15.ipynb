```python
#Write a python program to perform text Classification using LSTM 
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer 
from tensorflow.keras.preprocessing.sequence import pad_sequences 
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout 
```

```python
# Step 1: Prepare news dataset 
news = [ 
    # Sports 
    "India won the cricket world cup", 
    "The football team won the match",
    "Virat scored a century in cricket", 
    "The player won a gold medal", 
    "The cricket team played very well", 
    "India defeated Australia in cricket",
    # Politics 
    "The government announced a new policy",
    "The prime minister addressed the nation",
    "Election campaigns started in India", 
    "The parliament passed a new bill",
    "Political leaders attended the meeting", 
    "The government introduced new laws",
    # Technology 
    "Artificial intelligence is changing technology", 
    "Scientists developed a new computer", 
    "Machine learning is used in modern applications", 
    "A new smartphone was launched", 
    "Robots are becoming more intelligent",
    "Technology is improving every day", 
    # Entertainment 
    "The actor starred in a new movie",
    "The film received many awards",
    "The singer released a new song",
    "The movie was very popular", 
    "The actress performed beautifully",
    "The television show became famous"] 
```

```python
# Step 2: Assign numerical labels
# 0 = Sports
# 1 = Politics
# 2 = Technology
# 3 = Entertainment
labels = np.array([0]*6 + [1]*6 + [2]*6 + [3]*6) 
```

```python
# Step 3: Tokenization 
tokenizer = Tokenizer(num_words=1000, oov_token="<OOV>") 
tokenizer.fit_on_texts(news)
sequences = tokenizer.texts_to_sequences(news) 
```

```python
# Step 4: Padding
x = pad_sequences(sequences, maxlen=10, padding='post') 
y = labels 
```

```python
# Step 5: Create LSTM model
model = Sequential()
model.add(Embedding(input_dim=1000, output_dim=16, input_length=10))
model.add(LSTM(32))
model.add(Dropout(0.2))
model.add(Dense(4, activation='softmax')) 
```

```python
# Step 6: Compile the model 
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',
              metrics=['accuracy']) 
```

```python
# Step 7: Train the model
model.fit(x, y, epochs=50, verbose=1) 
```

```python
# Step 8: Test with new news articles
test_news = [ "India won the cricket match", 
             "The government passed a new law", 
             "Artificial intelligence is growing rapidly", 
             "The actor released a new movie"] 
```

```python
# Step 9: Convert test text into numerical values 
test_sequences = tokenizer.texts_to_sequences(test_news)
test_data = pad_sequences(test_sequences, maxlen=10, padding='post') 
```

```python
# Step 10: Predict news categories 
predictions = model.predict(test_data)
categories = ['Sports', 'Politics', 'Technology', 'Entertainment'] 
```

```python
# Step 11: Display results 
for i in range(len(test_news)): 
    predicted_class = np.argmax(predictions[i]) 
    print("News:", test_news[i]) 
    print("Predicted Category:", categories[predicted_class]) 
    print() 
```

```python
import torch 
import torch.nn as nn
import torch.optim as optim 
```

```python
# Training data
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]]) 
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]]) 
```

```python
# Create neural network 
model = nn.Sequential( nn.Linear(1, 10), nn.ReLU(), nn.Linear(10, 1)) 
```

```python
# Loss function 
criterion = nn.MSELoss() 
```

```python
# Optimizer
optimizer = optim.SGD(model.parameters(), lr=0.01) 
# Training
for epoch in range(1000): 
    # Forward propagation
    output = model(X) 
# Calculate loss 
    loss = criterion(output, y) 
# Backpropagation 
    optimizer.zero_grad() 
    loss.backward() 
# Update weights 
    optimizer.step() 
# Prediction
test = torch.tensor([[12.0]]) 
prediction = model(test)
print("Prediction:", round(prediction.item()) )
```
