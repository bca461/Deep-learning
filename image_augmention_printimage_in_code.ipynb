```python
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
image_path="C:\\Users\\BCA LAB 10\\Desktop\\Ishika\\laptopjpg"
img=tf.keras.utils.load_img("laptop.jpg")
img_array=tf.keras.utils.img_to_array(img)
img_array=img_array.reshape((1,)+img_array.shape)
```

```python
augmentations={"Original":ImageDataGenerator(),
               "Rotation":ImageDataGenerator(rotation_range=50),
               "width Shift":ImageDataGenerator(width_shift_range=0.5),
               "Height Shift":ImageDataGenerator(height_shift_range=0.5),
               "Zoom":ImageDataGenerator(zoom_range=0.5),
               "Vertical Flip":ImageDataGenerator(vertical_flip=True),
               "Shear":ImageDataGenerator(shear_range=50)}
```

```python
plt.figure(figsize=(12,8))
for i,(name,datagen) in enumerate(augmentations.items()):
    if name=="Original":
        augmented_image=img_array[0]
    else:
        batch=next(datagen.flow(img_array,batch_size=1,shuffle=False))
        augmented_image=batch[0]
        plt.subplot(2,4,i+1)
        plt.imshow(augmented_image.astype("uint8"))
        plt.title(name)
        plt.axis("off")
        plt.tight_layout()
        plt.show()

    
```
