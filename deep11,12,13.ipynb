```python
import numpy as np 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
# Generate training data
data = np.arange(0, 1000, 10, dtype=float)
# Normalize data
data = data / 100.0
x, y = [], [] 
for i in range(len(data) - 3):
    x.append(data[i:i+3]) 
    y.append(data[i+3]) 
    
x = np.array(x).reshape(-1, 3, 1)
y = np.array(y)
# Build RNN model
model = Sequential([SimpleRNN(32, activation='tanh', input_shape=(3, 1)),Dense(1)]) 
# Compile model 
model.compile(optimizer='adam', loss='mse')
# Train model 
model.fit(x, y, epochs=300, batch_size=8, verbose=0)
# Test with a new sequence
test = np.array([70, 80, 90], dtype=float) / 100.0 
test = test.reshape(1, 3, 1) 
# Predict next number
prediction = model.predict(test, verbose=0)[0, 0] * 100.0
print("Input sequence:", [70, 80, 90])
print("Predicted next number:", round(prediction, 2)) 
```

```python
#Write a Python program using TensorFlow/Keras to implement an LSTM (Long Short-Term Memory) 
#neural network for sequence prediction. 
import numpy as np
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import LSTM, Dense 
from tensorflow.keras.callbacks import EarlyStopping 
# Generate sequence data
data = np.arange(0, 1000, 10, dtype=float) 
# Normalize data
data = data / 1000.0 
x, y = [], [] 
for i in range(len(data) - 3): 
    x.append(data[i:i+3])
    y.append(data[i+3])
x = np.array(x).reshape(-1, 3, 1)
y = np.array(y) 
# Create model
model = Sequential([LSTM(32, input_shape=(3, 1)), Dense(1)]) 
# Compile model 
model.compile(optimizer='adam', loss='mse') 
# Train model 
model.fit(x, y,epochs=500,batch_size=8,verbose=0, callbacks=[EarlyStopping(monitor='loss',patience=30, restore_best_weights=True)])
# Predict 
test = np.array([60, 70, 80], dtype=float)
test = (test / 1000.0).reshape(1, 3, 1) 
prediction = model.predict(test, verbose=0)[0, 0] * 1000
print("Input sequence: 60, 70, 80")
print("Predicted next number:", round(prediction, 2)) 
```

```python
#Write a Python program to perform Sentiment Analysis using an LSTM model. 
import numpy as np 
import tensorflow as tf 
from tensorflow.keras.datasets import imdb 
from tensorflow.keras.preprocessing.sequence import pad_sequences 
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout 
# 1. Load IMDB dataset 
vocab_size=10000
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000) 
# 2. Apply Padding
x_train = pad_sequences(x_train, maxlen=200)
x_test = pad_sequences(x_test, maxlen=200) 
# 3. Build LSTM Model 
model = Sequential()
model.add(Embedding(vocab_size, 128,input_length=200)) 
model.add(LSTM(64)) 
model.add(Dropout(0.5)) 
model.add(Dense(1, activation='sigmoid'))
# 4. Compile Model 
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
# 5. Train Model 
model.fit(x_train, y_train, epochs=3, batch_size=64, validation_split=0.2) 
# 6. Evaluate Model 
loss, accuracy = model.evaluate(x_test, y_test) 
print("Test Accuracy:", accuracy) 
# 7. Predict Sentiment
prediction = model.predict(x_test[:1])
if prediction[0][0] >= 0.5: 
    print("Positive Sentiment") 
else: 
    print("Negative Sentiment") 
```
