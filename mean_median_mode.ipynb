```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame({"Marks":[20,30,None,40,None]})
df
```

```python
df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
df
```

```python
ndf=pd.DataFrame({"Score":[10,30,None,15,None]})
ndf
```

```python
ndf["Score"]=ndf["Score"].fillna(ndf["Score"].median())
ndf
```

```python
a=pd.DataFrame({"RUN":[20,10,None,10,15,None]})
a
```

```python
a["RUN"]=a["RUN"].fillna(a["RUN"].mode()[0])
a
```

```python
b=pd.DataFrame({"NAME":["ishika","puja",None,"puja","piku",None]})
b
```

```python
b["NAME"]=b["NAME"].fillna(b["NAME"].mode()[0])
b
```
